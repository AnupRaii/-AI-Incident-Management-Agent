```python
from crewai.tools import tool

@tool("Get Service Status")
def get_service_status(service_name: str) -> str:
    """
    Retrieves the current operational status
    of a university IT service.
    """

    status = service_status.get(service_name)

    if status is None:
        return f"Service {service_name} not found."

    return f"{service_name} status is {status}."
```

```python
result = get_service_status.run("ERP")
print(result)

result = get_service_status.run("Payment Gateway")
print(result)
```

```python
@tool("Search System Logs")
def search_system_logs(service_name: str) -> str:
    """
    Searches system logs for a specific university IT service.
    Returns all matching log entries.
    """

    matching_logs = [
        entry["log"]
        for entry in system_logs
        if entry["service"].lower() == service_name.lower()
    ]

    if not matching_logs:
        return f"No logs found for {service_name}."

    return "\n".join(matching_logs)
```

```python
print("ERP LOGS:")
print(search_system_logs.run("ERP"))

print("\nPAYMENT GATEWAY LOGS:")
print(search_system_logs.run("Payment Gateway"))

print("\nLMS LOGS:")
print(search_system_logs.run("LMS"))
```

```python
@tool("Get Incident History")
def get_incident_history(service_name: str) -> str:
    """
    Retrieves available incident records for a specific service.
    """

    matching_incidents = [
        incident
        for incident in incident_reports
        if incident["service"].lower() == service_name.lower()
    ]

    if not matching_incidents:
        return f"No incident records found for {service_name}."

    results = []

    for incident in matching_incidents:
        results.append(
            f"Incident ID: {incident['id']}\n"
            f"Service: {incident['service']}\n"
            f"Description: {incident['description']}\n"
            f"Affected Users: {incident['affected_users']}"
        )

    return "\n\n".join(results)
```

```python
print("ERP INCIDENT HISTORY:\n")

print(get_incident_history.run("ERP"))

print("\nPAYMENT GATEWAY INCIDENT HISTORY:\n")

print(get_incident_history.run("Payment Gateway"))
```

```python
@tool("Calculate Incident Severity")
def calculate_incident_severity(
    service_name: str,
    affected_users: int
) -> str:
    """
    Calculates incident severity using service status
    and number of affected users.
    """

    status = service_status.get(service_name)

    if status is None:
        return f"Unknown service: {service_name}"

    if status == "Down":
        severity = "Critical"

    elif status == "Degraded" and affected_users >= 200:
        severity = "High"

    elif status == "Slow" and affected_users >= 100:
        severity = "High"

    elif status == "Degraded" or affected_users >= 50:
        severity = "Medium"

    else:
        severity = "Low"

    return (
        f"Incident Severity: {severity}\n"
        f"Service: {service_name}\n"
        f"Status: {status}\n"
        f"Affected Users: {affected_users}"
    )
```

```python
triage_agent = Agent(
    role="IT Incident Triage Specialist",

    goal=(
        "Analyze incoming university IT incidents, "
        "identify affected services and users, "
        "and classify incidents as Low, Medium, "
        "High or Critical."
    ),

    backstory=(
        "You are an experienced IT service desk specialist. "
        "You specialize in incident classification, "
        "business impact analysis and prioritization."
    ),

    tools=[
        get_service_status,
        calculate_incident_severity
    ],

    llm=llm,
    verbose=True,
    allow_delegation=False
)

print("Incident Triage Agent created!")
```

```python
investigation_agent = Agent(
    role="IT Technical Investigation Specialist",

    goal=(
        "Investigate university IT incidents using "
        "system logs, service status and incident records. "
        "Identify possible root causes and correlations."
    ),

    backstory=(
        "You are a senior system administrator with expertise "
        "in database troubleshooting, authentication systems, "
        "API failures and server performance monitoring."
    ),

    tools=[
        get_service_status,
        search_system_logs,
        get_incident_history
    ],

    llm=llm,
    verbose=True,
    allow_delegation=False
)

print("Technical Investigation Agent created!")
```

```python
response_agent = Agent(
    role="IT Incident Response Manager",

    goal=(
        "Prepare recovery plans, recommend corrective actions, "
        "prioritize incidents and generate a structured "
        "incident response report."
    ),

    backstory=(
        "You are an experienced IT operations manager "
        "responsible for maintaining university IT services "
        "and coordinating incident recovery."
    ),

    llm=llm,
    verbose=True,
    allow_delegation=False
)

print("Incident Response Agent created!")
```

```python
reviewer_agent = Agent(
    role="IT Incident Quality Reviewer",

    goal=(
        "Review incident investigation reports and verify "
        "that root causes, severity classifications and "
        "recommendations are supported by evidence."
    ),

    backstory=(
        "You are a senior IT quality auditor. "
        "You carefully inspect technical evidence, "
        "identify unsupported assumptions and ensure "
        "incident reports are accurate and complete."
    ),

    llm=llm,
    verbose=True,
    allow_delegation=False
)

print("Incident Quality Reviewer created!")
```

```python
agents = [
    triage_agent,
    investigation_agent,
    response_agent,
    reviewer_agent
]

for agent in agents:
    print("Agent:", agent.role)
```

```python
from crewai import Task

triage_task = Task(
    description="""
    Analyze all university IT incident reports.

    Incident data:
    {incident_data}

    For each incident:
    1. Identify the affected service.
    2. Identify the number of affected users.
    3. Check the service status using your tools.
    4. Calculate incident severity using your tool.
    5. Explain the business impact.

    Classify incidents as Low, Medium, High or Critical.

    Do not invent information.
    """,

    expected_output="""
    A structured triage report containing:
    Incident ID, affected service, affected users,
    service status, severity and business impact.
    """,

    agent=triage_agent
)

print("Triage task created!")
```

```python
investigation_task = Task(
    description="""
    Investigate the incidents using the triage findings.

    For each incident:
    1. Search relevant system logs.
    2. Check current service availability.
    3. Retrieve incident history.
    4. Identify possible technical root causes.
    5. Identify any common symptoms or dependencies.

    Clearly distinguish evidence from hypotheses.
    Do not invent technical findings.
    """,

    expected_output="""
    A detailed technical investigation report containing
    incident ID, log evidence, service status,
    possible root cause and incident correlations.
    """,

    agent=investigation_agent,
    context=[triage_task]
)

print("Investigation task created!")
```

```python
response_task = Task(
    description="""
    Prepare a recovery plan using the triage and
    technical investigation findings.

    For each incident:
    1. Recommend corrective actions.
    2. Explain why the actions are appropriate.
    3. Prioritize incidents by severity.
    4. Suggest recovery and verification steps.

    Do not execute any recovery action.

    Generate a structured incident response report.
    """,

    expected_output="""
    An incident response report containing:
    Executive summary, incident priorities,
    recommended recovery actions and verification steps.
    """,

    agent=response_agent,
    context=[triage_task, investigation_task]
)

print("Response task created!")
```

```python
review_task = Task(
    description="""
    Review the incident response report.

    Verify:
    1. Incident severity is supported by the data.
    2. Root causes are supported by system logs.
    3. Recommendations address identified issues.
    4. No unsupported claims are included.
    5. Human approval status is correctly represented.

    Identify missing evidence and correct errors.

    Produce the final verified incident report.
    """,

    expected_output="""
    A verified final incident report containing:
    Executive summary, incident findings,
    root causes, severity, recovery recommendations,
    quality review findings and approval status.
    """,

    agent=reviewer_agent,
    context=[triage_task, investigation_task, response_task],
    output_file="incident_report.md"
)

print("Quality review task created!")
```

```python
tasks = [
    triage_task,
    investigation_task,
    response_task,
    review_task
]

for i, task in enumerate(tasks, 1):
    print(f"Task {i}: {task.agent.role}")
```

```python
from crewai import Crew, Process
```

```python
incident_crew = Crew(

    agents=[
        triage_agent,
        investigation_agent,
        response_agent
    ],

    tasks=[
        triage_task,
        investigation_task,
        response_task
    ],

    process=Process.sequential,

    verbose=True
)

print("Incident Management Crew created successfully!")
```

```python
print("WORKFLOW: SEQUENTIAL")

for i, task in enumerate(incident_crew.tasks, 1):
    print(f"{i}. {task.agent.role}")
```

```python

# Execute CrewAI asynchronously in Google Colab

incident_result = await incident_crew.kickoff_async(
    inputs={
        "incident_data": json.dumps(
            incident_reports,
            indent=4
        )
    }
)

print("\n===== INCIDENT MANAGEMENT RESULTS =====\n")

print(incident_result)
```

```python
review_task = Task(
    description="""
    Review the following university IT incident report.

    PREVIOUS INVESTIGATION REPORT:
    {previous_report}

    HUMAN APPROVAL STATUS:
    {approval_status}

    Verify:
    1. Severity classifications are correct.
    2. Root causes are supported by system logs.
    3. Recommendations address identified problems.
    4. Unsupported assumptions are identified.
    5. Human approval status is accurately recorded.

    If approval is rejected, clearly state that
    no recovery action is authorized.

    Generate the final verified Markdown report.
    """,

    expected_output="""
    A complete verified incident report containing:
    Executive summary, incident details, evidence,
    root causes, severity, recovery recommendations,
    human approval status and quality review findings.
    """,

    agent=reviewer_agent,
    output_file="incident_report.md"
)

print("Quality Reviewer task configured!")
```

```python
reviewer_crew = Crew(
    agents=[reviewer_agent],

    tasks=[review_task],

    process=Process.sequential,

    verbose=True
)

print("Quality Reviewer Crew created successfully!")
```

```python
final_report = await reviewer_crew.kickoff_async(
    inputs={
        "previous_report": str(incident_result),
        "approval_status": approval_status
    }
)

print("\n===== FINAL VERIFIED INCIDENT REPORT =====\n")

print(final_report)
```
